
Ejercicio 12: Suma de pares del 1 al 20 con for y continue
Salida: La suma de los números pares del 1 al 20 es: 110

```python
numero = int(input("Dame un número"))
suma = 0
for n in range(numero + 1):
    if n % 2 == 0:
        suma = suma + n
    else:
        continue
print(f"La suma de los números pares es: {suma}")
```

Ejercicio 13: Suma de una lista con función que regresa valor (sin sum())
Entrada: [10, 20, 30, 40] → Salida: La suma de los elementos es: 100

```python
def suma_lista(lista):
    resultado = 0
    for n in lista:
        resultado = resultado + n
    return resultado

lista = [10, 20, 30, 40]
print(f"La suma de los elementos es: {suma_lista(lista)}")
```

Ejercicio 14: Máximo de una lista con función que regresa valor (sin max())
Entrada: [7, 3, 9, 2] → Salida: El elemento máximo es: 9

```python
def suma_lista(lista):
    lista.sort(reverse=True)
    return lista[0]

lista = [7, 3, 9, 2,]
print(f"El número maximo es: {suma_lista(lista)}")
```

Ejercicio 19: Tupla: acceso y suma con función
Entrada: (4, 8, 15, 16, 23, 42) → Salida:

El tercer elemento es: 15
La suma de la tupla es: 108

```python
def suma_tupla(tupla):
    resultado = 0
    for n in tupla:
        resultado = resultado + n
    return resultado
def tercer_elem(tupla):
    return tupla[2]

tupla = (4, 8, 15, 16, 23, 42)
print(f"El tercer elemento es: {tercer_elem(tupla)}", end=" ")
print(f"La suma de la tupla es: {suma_tupla(tupla)}")
```

Ejercicio 20: Diccionario: buscar teléfono de un contacto
Entrada: Beto → Salida: El teléfono de Beto es: 5555678

```python
def buscar_contacto(contactos, nombre):
    return contactos.get(nombre, "Contacto no encontrado")
     
contactos = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103",
    "Beto": "5555-678"
}
nombre = input(print("Escribe el nombre del contacto que buscas"))
tel = buscar_contacto(contactos, nombre.capitalize())
print(f"El número de {nombre.capitalize()} es: {tel}")
```

Ejercicio 22: Excepción de división entre cero
Entrada: 10, 0 → Salida: Error: No se puede dividir entre cero.

```python
try:
   dividendo = int(input("Dame un número"))
   divisor = int(input("Dame otro número"))
except ValueError:
   print("Datos incorrectos\nIngresa números enteros")
else:
    try:
        resultado = dividendo / divisor
        print(f"{dividendo} ÷ {divisor} = {resultado}")
    except ZeroDivisionError:
       print("Error. No se puede dividir entre cero")
```
