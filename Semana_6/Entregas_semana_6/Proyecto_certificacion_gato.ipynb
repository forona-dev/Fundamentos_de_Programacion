Escenario
Tu tarea es escribir un simple programa que simule jugar a tic-tac-toe (nombre en inglés) con el usuario. Para hacerlo más fácil, hemos decidido simplificar el juego. Aquí están nuestras reglas:

la maquina (por ejemplo, el programa) jugará utilizando las 'X's;
el usuario (por ejemplo, tu) jugarás utilizando las 'O's;
el primer movimiento es de la maquina - siempre coloca una 'X' en el centro del tablero;
todos los cuadros están numerados comenzando con el 1 (observa el ejemplo para que tengas una referencia)
el usuario ingresa su movimiento introduciendo el número de cuadro elegido - el número debe de ser valido, por ejemplo un valor entero mayor que 0 y menor que 10, y no puede ser un cuadro que ya esté ocupado;
el programa verifica si el juego ha terminado - existen cuatro posibles veredictos: el juego continua, el juego termina en empate, tu ganas, o la maquina gana;
la maquina responde con su movimiento y se verifica el estado del juego;
no se debe implementar algún tipo de inteligencia artificial - la maquina elegirá un cuadro de manera aleatoria, eso es suficiente para este juego.

```python
def display_board(board):
    for row in board:                           # Este for recorre las filas del tablero
        print("+-------+-------+-------+")      # Imprime el marco superior de cada fila
        print("|       |       |       |")      # Imprime los marcos laterales de las filas (arriba del número)
        for column in row:                      # Este for recorre las columnas de cada fila
            print(f"|   {column}   ", end="")   # Imprime el número asignado a cada casilla
        print("|")                              # Imprime la barra a la derecha del número de la ultima columna
        print("|       |       |       |")      # Imprime los marcos laterales de las filas (debajo del número )
    print("+-------+-------+-------+")          # Imprime el marco inferior del tablero



def enter_move(board):
    ent = 1                                                  # Variable de control del bucle de la función
    while ent == 1:
        try:
            casilla = int(input("Escoge una casilla"))       # Pedimos al usuario que escoja una casilla para su turno
        except ValueError:                                   # Verificamos que el dato ingresado sea correcto
            print("Dato incorrecto")
            continue                                         # Pasamos a la siguiente iteración del ciclo (permite volver a pedir una casilla en caso de que se ingrese un dato incorrecto)

        if casilla <1 or casilla >9:                         # Verificamos que el dato ingresado este dentro del rango
            print("Dato fuera de rango")
            continue                                         # Pasamos a la siguiente iteración del ciclo (permite volver a pedir una casilla en caso de que se ingrese un dato incorrecto)

                                                             # Ahora hay que convertir la casilla seleccionada a coordenadas de la matriz
        row = (casilla -1) // 3                              # Para las filas iniciamos restando 1 al valor de la casilla (esto porque Python empieza a contar desde el 0) despues realizamos
                                                             # una divicion entera (//) entre 3 (el 3 representa el número de filas del tablero). El resultado equivale al numero de fila (0, 1, 2)
        column = (casilla -1) % 3                            # Para las columnas usamos el residuo de la división entre 3, para eso usamos el operador %. AL combinar ambos resultados obtenemos
                                                             # las coordenadas de la matriz que equivalen a ese número de casilla

        if (row, column) in make_list_of_free_fields(board): # Aqui verificamos que las coordenadas de la casilla seleccionada esten en la lista de casillas disponubles con "in" y llamamos a la funcion que contiene la lista
            board[row][column] = "O"                         # Si las coordenadas estan en la lista se asigna "O" al valor de dicha coordenada
            ent = 0                                          # Cambiamos el valor de la variable de control para romper el ciclo
        else: 
            print("Esa casilla ya esta ocupada")             # Si las coordenadas no estan en la lista, avisamos al usuario que esa casilla ya fue seleccionada



def make_list_of_free_fields(board):
    cas_disp = []                                           # Creamos una lista que contendra las casillas disponibles
    for r in range(3):                                      # Este for recorre las filas del tablero (board)
        for c in range(3):                                  # Este for recorre las columnas del tableo (board)
            if board[r][c] != "X" and board[r][c] != "O":   # Aqui revisamos que el dato contenido en la casilla (fila y columna) sea distinto de "X" y de "O"
                cas_disp.append((r,c))                      # Si el dato contenido no es "X" ni "O" se guardan los valores de la fila (r) y de la columna (c) como tupla y se agregan a la lista con .append()
    return cas_disp                                         # La función retorna la lista con las tuplas que representan las coordenadas (fila, columna) de cada casilla disponible en el tablero 


def victory_for(board, sign):
    for r in range(3):                                                                  # Aqui recorremos las filas
        if (board[r][0] == sign) and (board[r][1] == sign) and (board[r][2] == sign):   # Aqui verificamos si hay tres signos iguales en una misma fila
            return True                                                                 # Si se encuentran 3 signos iguales en una fila el resultado es verdadero
                                                                                        # (La asignación de "X" y "O" al signo se hará dentro del bucle del juego)
    for c in range(3):                                                                  # Aqui recorremos las columnas
        if (board[0][c] == sign) and (board[1][c] == sign) and (board[2][c] == sign):   # Aqui verificamos si hay tres signos iguales en la misma columna
            return True                                                                 # Si se encuentran 3 signos iguales en una columna el resultado de la función es verdadero
    if (board[0][0] == sign and board[1][1] == sign and board[2][2] == sign):           # Aqui verificamos si hay tres signos iguales en la diagonal de izq-der
        return True                                                                     # Si se encuenta el mismo signo en la diagonal izq-der el resultado es verdadero
    if (board[0][2] == sign and board[1][1] == sign and board[2][0] == sign):           # Aqui verificamos si hay tres signos iguales en la diagonal de der-izq
        return True                                                                     # Si se encuentran 3 signos iguales en la diagonal der-izq el resultado es verdadero
    return False                                                                        # Si ninguna de las condiciones resulta verdadera la funcion retorna falso y el juego continua 


from random import randrange
def draw_move(board):
    cas_lib = make_list_of_free_fields(board)   # creamos una variable que contenga una lista de las casillas libres del tablero pasandole la funcion que hace la lista de casillas 
                                                # libres (de esta forma cada que se ocupa una casilla la lista guardada en la variable se actualiza junto con la función). Asi la  
                                                # computadora nunca escogerá una casilla ocupada, porque estas no forman parte de la lista
    turno_comp = randrange(len(cas_lib))        # Aqui usamos randrange para generar el turno de la computadora. usamos len para seleccionar un número al azar del 0 al la cantidad
                                                # de elementos en la lista -1 (randrange funciona similar a range (si hay 8 elementos en la lista randrange genera un número aleatorio del 0-7))
    row, column = cas_lib[turno_comp]           # tomamos el número aleatoreo y lo pasamos como indice de la lista. Recordemos que la lista contiene las tuplas con las coordenadas de
                                                # las casillas libres, por ejemplo si solo se ha seleccionado la casilla 5 la lista se veria asi: [(0, 0), (0, 1), (0, 2), (1, 0), (1, 2), (2, 0), (2, 1), (2, 2)]
                                                # si el número generado por rand range es el 2 de modo que row, column = cas_lib[2], entonces el programa busca lo que esta contenido en el 
                                                # indice 2 de la lista que seria la tupla (0, 2) que equivale a la casilla 3
    board[row][column] = "X"                    # Actualizamos el tablero marcando xon una "X" la casilla selecionada


board = [           # Esta es la matriz que contiene al tablero
    [1, 2, 3],
   [4, "X", 6],     # El tablero inicia con la casilla 5 marcada con el primer movimiento de la computadora "X"
    [7, 8, 9]
]

while True:
    display_board(board)                                            # Llamamos a la función que despliega el tablero
    enter_move(board)                                               # Llamamos a la función que pide al usuario que escoja una casilla
    victory_for(board, "O")                                         # Revisamos si el usuario ganó con la función de victoria (aqui hacemos la asignacion del signo del usuario, dandole como argumento "O" a la función)
    if victory_for(board, "O") == True:                             # Si la función retorna verdadero el usuario gana
        print("VICTORIA!!!!\nFlelicidades, haz gando!!! :)")
        display_board(board)                                        # Mostramos nuevamente el tablero para visualisar el resultado final de la partida si gana el usuario
        break                                                       # Usamos la sentencia break para romper el ciclo si gana el usuario                                                        
    draw_move(board)                                                # Llamamos a la función que genera el movimiento de la computadora
    victory_for(board, "X")                                         # Revisamos si la computadora ganó con la función de victoria (aqui hacemos la asignacion del signo de la computadora, dandole como argumento "X" a la función)
    if victory_for(board, "X") == True:                             # Si la función retorna verdadero la computadora gana
        print("OH NO!!! Haz perdido\nMejor suerte a la próxima")
        display_board(board)                                        # Mostramos nuevamente el tablero para visualisar el resultado final de la partida si gana la computadora
        break                                                       # Usamos la sentencia break para romper el ciclo si gana la computadora                             
    if make_list_of_free_fields(board) == []:                       # Revisamos si la lista está vacia, si ya no quedan más casillas disponibles el juego termina en empate
        print("No quedan más casillas disponibles\nEs un empate!")  
        display_board(board)                                        # Mostramos nuevamente el tablero para visualisar el resultado final de la partida si hay un empate
        break                                                       # Usamos la sentencia break para romper el ciclo si hay un empate
```
