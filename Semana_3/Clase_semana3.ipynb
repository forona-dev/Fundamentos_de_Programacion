```python
divisor = int(input("Ingresa un número entero y positivo"))
div = divisor
dividendo = int(input("Ingresa otro número entero y positivo"))
contador = 0
cociente = div - dividendo

while div >=0:
    cociente = div - dividendo
    contador = contador +1
    div = cociente
    print(f"cociente {cociente}")
    print(f"contador {contador}")
    print(f"div      {div}")
    if div <=0:
        print(f"{divisor} ÷ {dividendo} = {contador} con residuo: {div}")
        break
```

Enunciado: Crea un programa que pida un entero positivo n y calcule la suma alternada 1 - 2 + 3 - 4 + ... hasta n. Usa un for y decide el signo de cada término con el módulo %. Entrada: 6 Proceso: 1 - 2 + 3 - 4 + 5 - 6 = -3 Salida: El resultado de la serie es: -3

```python
numero = int(input("Ingresa un número entero y positivo"))
resultado = 0
for n in range(1, numero+1):
    if n % 2 == 0:
        resultado = resultado + n
    else:
        resultado = resultado - n
print(f"El resultado de la serie es: {resultado}")
```
