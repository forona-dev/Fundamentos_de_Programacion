Extra 1: Sistema de calificaciones con tuplas
Enunciado: Crea una tupla calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0). Imprime la tercera calificación, captura dos calificaciones nuevas con input() y anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala de mayor a menor y muestra la suma de todos los elementos usando una función que retorne el valor. Entrada:

Nueva calificación 1: 8.5
Nueva calificación 2: 9.5

Salida:

Tercera calificación: 8.0
Nueva tupla: (7.5, 9.0, 8.0, 6.5, 10.0, 8.5, 9.5)
Lista ordenada (mayor a menor): [10.0, 9.5, 9.0, 8.5, 8.0, 7.5, 6.5]
Suma total: 59.0

```python
def suma_calif(lista_calif):
    total = 0
    for c in lista_calif:
        total = total + c
    return total
var = 1
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)
print(f"Tupla: {calificaciones}")
print(f"Tercera calificación: {calificaciones[2]}")
while var == 1:
    try:
        calif_1 = float(input("Ingresa una nueva calificación"))
        calif_2 = float(input("Ingresa una nueva calificación"))
    except ValueError:
        print("valir invalido")
    else:
        calif_new = calificaciones + (calif_1, calif_2)
        lista_calif = list(calif_new)
        lista_calif.sort(reverse=True)
        print(f"Nueva tupla: {calif_new}")
        print(f"Calificaciones ordenadas: {lista_calif}")
        print(f"La suma de todas las calificaciones es: {suma_calif(lista_calif)}")
        var = 0
```

Extra 2: Agenda de contactos con búsqueda
Enunciado: Crea un diccionario agenda con tres contactos iniciales ("Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103"). Captura un nuevo contacto (nombre y teléfono) con input() y agrégalo al diccionario. Luego imprime todos los nombres con keys() en una sola línea y, con una función que retorne el valor, busca y muestra el teléfono del nuevo contacto. Entrada:

Nombre del nuevo contacto: Pedro
Teléfono del nuevo contacto: 555-0140
Nombre a buscar: Pedro

Salida:

Contactos registrados: Ana, Luis, Mía, Pedro
El teléfono de Pedro es: 555-0140

```python
def buscar_con(agenda, nombre):
    return agenda.get(nombre, "Lo siento, no encontre el contacto :(")


agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}
new_con = input("nombre del nuevo contacto")
new_num = input("Teléfono del nuevo contacto: ")
agenda [new_con.capitalize()] = new_num
print("Contactos registrados: ", end="")
for n in agenda.keys():
    print(n, end=", ")

nombre = input("Escribe el nombre del contacto que buscas")
num_tel = buscar_con(agenda, nombre.capitalize())
print(f"\nEl número de {nombre.capitalize()} es: {num_tel}") 
```

Extra 3: Calculadora segura con manejo de excepciones
Enunciado: Escribe un programa que pida dos números enteros. Implementa un bloque try-except que capture ValueError (caracteres no numéricos o vacíos) con un mensaje controlado y un bloque específico except ZeroDivisionError que muestre un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la división del primero entre el segundo. Entrada:

Primer número entero: 20
Segundo número entero: 4

Salida:

20 dividido entre 4 es: 5.0

Caso de error (divisor cero):

Primer número entero: 10
Segundo número entero: 0

Salida:

Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.

```python
var = 1
while var ==1:
    try:
        dividendo = int(input("Ingresa un número entero para el dividendo"))
        divisor = int(input("Ingresa otro número entero para el divisor"))
        break
    except ValueError:
        print("Parece que ingresaste un dato invalido.\nIntenta de nuevo ingresando solo números enteros ;)\n")
try:
    resultado = dividendo / divisor
    print(f"{dividendo} dividido entre {divisor} es: {resultado}")
    var = 0
except ZeroDivisionError:
    print("No es posible dividir entre cero.\nIngresa un divisor distinto de 0.")
    var = 0
```

Extra 4: Analizador de mensajes con strings
Enunciado: Crea una variable mensaje = "Python es un lenguaje poderoso". Imprime su longitud con len(), conviértelo a mayúsculas, reemplaza la palabra "Python" por "programación" y, con una función que retorne el conteo, muestra cuántas palabras contiene el mensaje original. Entrada:

mensaje = "Python es un lenguaje poderoso"

Salida:

Longitud del mensaje: 30
En mayúsculas: PYTHON ES UN LENGUAJE PODEROSO
Texto reemplazado: programación es un lenguaje poderoso
Palabras totales: 5

```python
def cont_palabras(mensaje):
    palabras = mensaje.split()
    return len(palabras)

mensaje = "Python es un lenguaje poderoso"
len_msj = len(mensaje)
msj_upper = mensaje.upper()
py_rplc = mensaje.replace("Python", "Programación")

print(f"La longitud del mensaje es de: {len_msj} caracteres")
print(f"Conversión a mayúsculas: {msj_upper}")
print(f"Texto reemplazado: {py_rplc}")
print(f"El mensaje tiene: {cont_palabras(mensaje)} palabras")
```

Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario votos = {"Rojo": 0, "Azul": 0, "Verde": 0}. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con items() el color ganador junto con su total.

```python
opciones = {
    "Rojo": 0,
    "Azul": 0,
    "Verde": 0
}
conteo = 1
while conteo <= 5:
        voto = input(f"Ingresa el voto: {conteo} (Rojo, Azul o Verde)")
        if voto.capitalize() in opciones:
            print(f"Voto valido para {voto.capitalize()}")
            opciones[voto.capitalize()] = opciones[voto.capitalize()] +1
            conteo = conteo +1
        else:
            print("El voto no es valido")

print("\nResultados")
for k, v in opciones.items():
     print(f"{v} votos para {k}") 
```

Desafío Extra 4: Analizador de textos
Crea un programa que reciba una frase, imprima su longitud con len(), la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.

```python
def cont_palabras(frase):
    palabras = frase.split()
    return len(palabras)

frase = input("Escribe una palabra o frase")
frase_length = len(frase)
frase_mayus = frase.upper()
python = "Python"
frase_switch = python.replace(python, "programación")

print(f"Tu frase tiene {frase_length} caracteres")
print(f"Convertir a mayúsculs: {frase} -> {frase_mayus}")
print(f"Convertir una palabra a otra: {python} -> {frase_switch}")
print(f"La frase tiene: {cont_palabras(frase)} palabra(s)")
```
