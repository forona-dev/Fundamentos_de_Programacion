TEMA 13: TUPLAS
Una tupla es una secuencia ordenada e inmutable de elementos en Python. A diferencia de las listas, una vez creada una tupla no se pueden agregar, eliminar ni modificar sus elementos. Se define con paréntesis ().

13.1 Creación de Tuplas


```python
# Tupla de números enteros
numeros = (10, 20, 30, 40, 50)

# Tupla de cadenas de texto
frutas = ("manzana", "plátano", "cereza")

# Tupla con tiopos mixtos
datos = ("Ana",  20, 1.65, True)

# Tupla de un solo elemento (obligatorio el parentesis y la coma)
unico = (5,)

print(type(numeros))
print(type(frutas))
print(type(datos))
print(type(unico))
```

```python
numeros = (10, 20, 30, 40, 50, 60, 70)

print(numeros[0])
print(numeros[1])
print(numeros[2])
print(numeros[3])
print(numeros[4])
print(numeros[-1])

#numeros[1] = 15
```

```python
coordenada = (2, 6)

x, y = coordenada

print(f"Coordenada en x: {x}")
print(f"Coordenada en y: {y}")
```

```python
def suma_tupla(t):
    total= 0
    for e in t:
        total = total + e
    return total

calorias = (800, 950, 700, 1300, 800, 900, 650)
resultado = suma_tupla(calorias)
print(f"Metiste a tu cuerpo {resultado} calorias, GORDO!") 
```

TEMA 14: DICCIONARIOS
Un diccionario es una colección de pares clave : valor encerrados entre llaves {}. Permite guardar datos asociados, por ejemplo, el nombre de un contacto (clave) con su número de teléfono (valor).

4.1 Claves y Valores
La clave es un identificador único e inmutable (generalmente un string o un número). El valor puede ser cualquier tipo de dato (número, string, lista, otro diccionario, etc.).

```python
contactos = {
    "Fili": "4423619012",
    "Ana": "44211144455",
    "Emi": "44233377799"
}

print(type(contactos))
print(len(contactos))
```

```python
contactos = {
    "Fili": "4423619012",
    "Ana": "44211144455",
    "Emi": "44233377799"
}

print(contactos["Fili"])
print(contactos.get("Raul"))
print(contactos.get("Raul", "Si registro"))
```

```python
contactos = {
    "Fili": "4423611528",
    "Ana": "44211144455",
    "Emi": "44233377799"
}

contactos ["Cris"] = "2299977755"
contactos ["Fili"] = "4423619012"

print(contactos)
```

```python
print(contactos.keys())
print(contactos.values())
print(contactos.items())
```

```python
print("Impresión de llaves")
for n in contactos.keys():
    print(n)

print("\nImpresión de valores")
for v in contactos.values():
    print(v)

print("\nImpresión completa")
for n, t in contactos.items():
    print(f"Nombre: {n}, Teléfono: {t}") 
```

```python
def buscar_contacto(contactos, nombre):
    return contactos.get(nombre, "Contacto no encontrado")

nombre = input("Escribe el nombre del contacto que buscas")
telefono = buscar_contacto(contactos, nombre)
print(f"El teléfono de {nombre} es {telefono}")
```

TEMA 15: EXCEPCIONES
Una excepción es un error que ocurre durante la ejecución del programa y que, si no se maneja, provoca que el programa se detenga. Con try/except podemos anticipar estos errores y desplegar mensajes controlados en lugar de crashear.

15.1 ¿Qué es una Excepción?
Ejemplos comunes de excepciones en Python:

ValueError: se intenta convertir texto no numérico con int() o float().
ZeroDivisionError: se intenta dividir un número entre cero.
TypeError: se opera con tipos incompatibles (ej. "Hola" + 5).
IndexError: se accede a un índice inexistente de una secuencia.

15.2 Estructura Básica try / except
El código que puede fallar se coloca dentro del bloque try. Si ocurre un error, el bloque except lo captura y ejecuta el código de recuperación.

try:
    edad = int(input("Ingresa tu edad: "))
    print("Edad registrada:", edad)
except ValueError:
    print("Error: Debes ingresar solo números enteros.")

Ejecución de ejemplo 1: Entrada 25 → Salida Edad registrada: 25
Ejecución de ejemplo 2: Entrada abc → Salida Error: Debes ingresar solo números enteros.

```python

## Excepciones
try:
    edad = int(input("Ingresa tu edad: "))
    print(f"Tu edad es: {edad}")
except ValueError:
    print("Error: Debes ingresar solo números enteros")
```

```python
try:
    dividendo = int(input("Dame un número: "))
    divisor = int(input("Entre que número lo quieres dividir"))
    resultado = dividendo / divisor
except ZeroDivisionError:
    print("Error de divición entre cero")
else:
    print(f"Division exitosa! El resultado es: {resultado}")
finally:
    print("Fue un gusto hacer esta divicion ;)")
```

```python
candidatos = {
    "A": 0,
    "B": 0,
    "C": 0
}

cantidad = int(input("Ccuantos votos se van a registrar"))
conteo = 1
while conteo <= cantidad:
        voto = input(f"Ingresa el voto: {conteo} (A, B o C)")
        if voto.upper() in candidatos:
            print(f"Voto valido para {voto.upper()}")
            candidatos[voto.upper()] = candidatos[voto.upper()] +1
            conteo = conteo +1
        else:
            print("El voto no es valido")

print("\nResultados")
for k, v in candidatos.items():
     print(f"Opcion {k} votos: {v}") 
```
