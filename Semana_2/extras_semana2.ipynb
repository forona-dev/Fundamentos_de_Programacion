Extra 1: Control de aforo con break y continue (while)
Enunciado: Crea un programa que procese visitantes del museo con un ciclo while. Por cada visitante se captura el costo de su boleto ($0 para menores de 3 años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis, muestra "Menor de 3 años, sin cargo" y continúa con el siguiente visitante (usa continue). Si el total acumulado alcanza o supera $100, detén el ciclo con break y muestra el total. Entrada:

Costo del boleto 1: 45
Costo del boleto 2: 0
Costo del boleto 3: 30
Costo del boleto 4: 45

Salida:

Menor de 3 años, sin cargo
Aforo alcanzado, se detiene el registro
Total acumulado: $120.00

```python
# Variables
visitante = 1               # Contador (va a ir contando a los visitantes)
boleto = 0                  # Acumulador (va a ir sumando el costo de los boletos)                 
precio = 0                  # Aqui se va a guarar el total de todos los boletos
while True:         
    boleto = int(input(f"costo del boleto {visitante}: "))
    precio = boleto + precio
    visitante = visitante +1
    if boleto == 0:
        print("Menor de 3 años, sin cargo")
    else:                   # Condocion para que el programa se detenga cuando la suma de los boletos alcance o supere "100"
        if precio >=100:
            break
print(f"Aforo alcanzado, se detiene el registro")
print(f"Total acumulado: ${precio}")                                           

```

Extra 2: Estadística de visitantes con for
Enunciado: Crea un programa que pida el número total de visitantes y, con un ciclo for, capture la edad de cada uno. Debe contar cuántos son adultos (18 años o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos resultados con dos decimales. Entrada:

Número de visitantes: 4
Edad del visitante 1: 30
Edad del visitante 2: 45
Edad del visitante 3: 15
Edad del visitante 4: 2

Salida:

Adultos: 2
Promedio de edad: 23.00

```python
# Variables
contador = 0                    # Va contando los visitantes
acumulador = 0                  # Va sumando las edades
adulto = 0
visitantes = int(input("Número de visitantes: ")) # Usuario ingresa el número de vicitantes (esto va a ser el numero de veces que se repetira el ciclo)
for contador in range(1, visitantes+1):
    edad = int(input(f"Edad del visitante: {contador}"))
    if edad >=18:               # Detecta si el visitante es mayor de edad
        adulto = adulto +1
    acumulador = acumulador + edad
promedio = acumulador / visitantes
print(f"Adultos: {adulto}")
print(f"Promedio de edad: {promedio:.2f} años")
```

Extra 3: Depuración de un cobro con PDB
Enunciado: El siguiente código pretende calcular el precio final de un boleto de $45 con el descuento del adulto mayor (12%) pero contiene un error de lógica. Inserta pdb.set_trace() para rastrear las variables, identifica el error y escribe la versión corregida.

precio = 45
descuento = 12
total = precio - descuento
print(f"Total: ${total:.2f}")

Entrada: Ninguna (código dado para depurar). Salida:

Error de lógica: descuento debe ser el porcentaje 0.12, no 12.
Código corregido: descuento = precio * 0.12 -> Total: $39.60

```python
import pdb; pdb.set_trace()

precio = 45
descuento = precio * 0.12
total = precio - descuento
print(f"Total: ${total:.2f}")
```

Extra 4: Pirámide de asteriscos con for
Enunciado: Crea un programa que pida la altura de una pirámide y la dibuje con asteriscos usando ciclos for anidados y el parámetro end de print. Entrada:

Altura de la pirámide: 4

Salida:

*
**
***
****

```python
# Variables
altura = int(input("Altura de la pirámide: "))
for a in range(1, altura +1):
    print()
    for b in range(1, a+1):
        print("*", end="")



```
