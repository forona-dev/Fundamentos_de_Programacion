### Ejercicio 11: Suma de los primeros N números
**Enunciado:** Crea un programa que pida un número entero positivo N y calcule con
`while` la suma de los números del 1 al N.
**Entrada:** `5`
**Salida:** `La suma del 1 al 5 es: 15`

````python
#Variables
total = 0
inicio = 1
#Selección del limite de la suma
N = int(input(print("Ingresa un número entero positivo")))
#Cálculo de la suma hasta el limite establecido por el usuario
while inicio <= N:
    total = total + inicio        # Acumulador (va sumando los valores)
    inicio = inicio + 1         # Sumador (evita que el acumulador caiga en un ciclo infinito)
#Mostrar resultados    
print(f"La suma del 1 al {N} es igual a {total}")
````

### Ejercicio 12: Tabla de multiplicar con while
**Enunciado:** Crea un programa que pida un número y muestre su tabla de
multiplicar del 1 al 10 usando `while`.
**Entrada:** `3`
**Salida:**
```
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
3 x 4 = 12
3 x 5 = 15
3 x 6 = 18
3 x 7 = 21
3 x 8 = 24
3 x 9 = 27
3 x 10 = 30

````python
#Variables
inicio = 1      #Este es mi contador (determina por qué número se multiplica "Tabla")
#Selección de la tabla de multiplicar
Tabla = int(input(print("¿Qué tabla quieres que te muestre?")))
print(f"¡Muy bien! Te mostraré la tabla del: {Tabla}")
#Cálculo de la tabla de multiplicar seleccionada por el usuario
while inicio <=10:
    Tabla * inicio           # Multiplicador
    print(f"{Tabla} X {inicio} = {Tabla * inicio}") #Muestra resultados en formato a x b = c
    inicio = inicio + 1      # Nota
````

Nota: Dentro del ciclo "while" tengo que poner el incremento del contador despues de la impresión del resultado. 
Si lo pongo antes el programa multiplica Tabla * inicio (pero el programa no guarda ni muestra este resultado), despues aumenta el contador en +1 (ahora el contador pasa de valer 1 a valer 2), y ahora si, el programa muestra la multiplicación pero como aumente el contador antes de mostrar el resultado la tabla inicia multiplicando *2

 # Lógica inicial (incorrecta)                                               
**paso 1: multiplicar**                                                     
**paso 2: aumentar el contador**                                            
**paso 3: mostrar multiplicación**                                          
# Lógica correcta
**paso 1: multiplicar**
**paso 2: mostrar multiplicación**
**paso 3. aumentar el contador**

````python
#Solucion con "for" del ejercicio 12

#Variables
Tabla = int(input(print("¿Qué tabla quieres que muestre?")))
#Establecer rango
for m in range(1, 11):
    print(f"{Tabla} X {m} = {Tabla * m}") #Muestra resultados en formato a x b = c
````

### Ejercicio 14: Salir del ciclo con break
**Enunciado:** Crea un programa que capture textos con `input()` dentro de un
`while True`. Si el usuario escribe `salir`, el ciclo se interrumpe con `break`.
**Entrada:** `hola`, `mundo`, `salir`
**Salida:**
```
hola

Fin del ciclo

````python
while True:                        # Se inicia el codigo con un "while True" (inicia el ciclo)
    texto = input()                # Variable (captura de texto)
    if texto == "salir":           # Establecer la condición que interrumpe el ciclo
        print("Fin del ciclo")     
        break                      
    else:                          
        print(f"{texto}")          # Mostraar resultados
    
````

### Ejercicio 18: Detener la suma con break
**Enunciado:** Crea un programa que sume números capturados con `input()`; si el
acumulado alcanza o supera 100, el ciclo se detiene con `break` y se muestra el
total.
**Entrada:** `50`, `30`, `25`
**Salida:** `Total: 105`

````python
#Variables
numero = 0
acumulador = 0
suma = 0
while acumulador >=0:
    numero = float(input())
    suma = numero + acumulador
    acumulador = acumulador + numero
    if acumulador >=100:
        break
print(f"Total: {suma}")
````
