```python
# codigo para calcular velocidad
#variables
Distancia = float(input(print("Ingresa la distancia en Km: ")))
Tiempo = float(input(print("Ingresa el tiempo en horas: " )))
#cálculo de velocidad
Velocidad = Distancia / Tiempo
#Impresión de datos
print("Distancia: ",Distancia, "km")
print("tiempo: ",Tiempo, "h")
print("Tú velocidad es de: ", Velocidad, "Km/h")
```
