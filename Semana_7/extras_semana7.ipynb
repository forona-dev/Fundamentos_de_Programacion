Desafío Extra 1: Registro de contactos
Crea un programa que pida el nombre y el teléfono de un contacto y los anexe al archivo contactos.txt con el formato nombre|telefono. Después lee el archivo y muestra todos los contactos registrados.

```python
archivo = open("contactos.txt", mode="w")
archivo.write("ana|555-1234\n")
archivo.write("luis|555-9876\n")
archivo.close()

with open("contactos.txt", "a") as archivo:
    nombre = input("Escribe el nombre del nuevo contacto")
    numero = input("Escribe el número del nuevo contacto")
    new_con = nombre + "|" + numero
    archivo.write(f"{new_con}\n")
    print("Contacto guardado")

with open("contactos.txt", "r") as archivo:
    print("Contactos guardados:")
    for linea in archivo:
        print(linea.strip())
    
```

Desafío Extra 2: Analizador de texto
Crea un programa que lea un archivo texto.txt y muestre cuántas líneas, cuántas palabras y cuántos caracteres (sin contar saltos de línea) contiene.

Pista: Cuenta las palabras con len(linea.split()) y los caracteres con len(linea.strip()).

```python
archivo = open("texto.txt", mode="w")
```

```python
with open("texto.txt", "r") as archivo:
    lineas = archivo.readlines()
    print(f"Total de lineas: {len(lineas)}")
    for linea in archivo:
        palabras = len(linea.split())
        caracteres = len(linea.strip())
    print(f"El archivo tiene {palabras} palabras")    
    print(f"El archivo tiene {caracteres} caracteres")
```
