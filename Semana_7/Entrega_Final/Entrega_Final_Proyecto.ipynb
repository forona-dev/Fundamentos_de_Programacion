```python
import time                                                     # Maus-herramienta que nos ayudará más adelante
import os                                                       # Importamos os, esta herramienta nos servira más adelante para verificar si un archivo existe en el disco duro 

def cargar_usuarios(archivo_usrs = "usuarios.txt"):                         # Función para crear un diccionario que contenga los datos del usuario y lo guarde en un archivo.txt. Asignamos el archivo usuarios.txt a la variable archivo_usrs para poder pasarselo a la funcion
    usuarios = {}                                                           # Creamos un diccionario vacío donde guardaremos usuario : contraseña
    if not os.path.exists(archivo_usrs):                                    # Revisamos si el archivo usuarios.txt no existe
        with open(archivo_usrs, mode="w") as archivo:                       # Aqui creamos el archivo en caso de que no existiera y se abre en modo write
            archivo.write("admin:1234\n")                                   # Aqui generamos una cuenta de administrador 

    with open(archivo_usrs, mode="r") as archivo:                           # Ya que creamos el archivo usuarios.txt lo abrimos en modo de lectura
        for linea in archivo:                                               # Con este for recorremos el archivo linea por linea
            linea = linea.strip()                                           # Usamos .strip() para seleccionar unicamente el texto contenido en la linea
            if linea and ":" in linea:                                      # Verificamos que la linea tenga el formato correcto
                usr, pswrd = linea.split(":", 1)                            # Aqui dividimos el texto de la linea en dos partes con .split(":", 1), lo que esta antes de los ':' se guarda en usr y lo que esta despues se guarda en pswrd
                usuarios[usr]=pswrd                                         # Aqui guardamos los datos en el diccionario (usuarios) el usuario (usr) como clave y la contraseña (pswrd) como valor
    return usuarios                                                         # la funcion retorna el diccionario con los datos que esten en usuarios.txt

def guardar_usr_nuevo(usr, pswrd, archivo_usrs = "usuarios.txt"):           # Funcion para guardar nuevos usuarios
    with open(archivo_usrs, mode="a") as archivo:                           # Abrimos usuarios.txt en modo append para ingresar nuevos datos sin perder los anteriores
        archivo.write(f"{usr}:{pswrd}\n")                                   # Agregamos el nuevo usuario y contraseña al archivo


def log_in():                                                                                 # Función para iniciar sesión
    usrs_existentes = cargar_usuarios()                                                       # Llamamos a la funcion cargar_usuarios y la asignamos a una variable  
    while True:                                                                               # Iniciamos un ciclo que atrapa al usuario en un bucle hasta que logre iniciar sesión
        print("Selecciona una opción:\n1: Iniciar sesión\n2: Registrar un nuevo usuario\n")   # Imprimimos un menú de inicio
        op_login = input("Seleciona una opción del 1-2")                                      # Pedimos al usuario que escoja una opción (1 o 2)

        if op_login == "1":                                                                   # Opción 1 inico de sesión
            print("Iniciar sesión\n")                                       
            usr = input("Ingresa tu nombre de usuario")                                       # Pedimos al usuario que ingrese el nombre con el que se registró
            pswrd = input("Ingresa tu contraseña")                                            # Pedimos al usuario que ingrese su contraseña
            if usr in usrs_existentes and usrs_existentes[usr] == pswrd:                      # Revisamos si el nombre ingresado existe en el diccionario y comprobamos si la contraseña ingresada coinside con la contraseña correspondiente a ese usuario
                return usr                                                                    # Si los dos pasos anteriores son verdaderos la funcion termina, retorna el nombre del usuario y se rompe el ciclo
            else:
                print("El usuario y/o la contraseña son incorrectos\n")                       # Si el usuario no existe o la contraseña no coinside mostramos un mensaje de error y regresamos al usuario al menú de inicio

        elif op_login =="2":                                                # Opción 2 registrar nueva cuenta
            print("Registrar nuevo usuario\n")
            usr_new = input("Crea un nombre de usuario")                    # Pedimos al usuario que cree un nombre para su cuenta
            if usr_new in usrs_existentes:                                  # Revisa si el nombre ingresado no fue ingresado previamente por otro usuario
                print("Ese nombre ya existe. Elige uno diferente\n")        # Mostramos un mensaje para informar al usuario que el nombre que eligio no esta disponible
            elif ":" in usr_new:                                            # Verificamos que el nombre no contenga ':' porque eso interferiria el formato de lectura del archivo (Usamos los ':' para separar el contenido de las lineas y poder guardar los datos en el diccionario con formato key:value)
                print("El nombre de usuario no puede contener ':'\n")       # Avisamos al usuario que no puede usar este caracter ':'
            else:
                pswrd_new = input("Crea una contraseña")                    # Si la creacion del nombre pasa todos los filtros le pedimos al usuario que cree una contraseña

                usrs_existentes[usr_new] = pswrd_new                        # Agregamos el nuevo usuario con su contraseña al diccionario
                guardar_usr_nuevo(usr_new, pswrd_new)                       # Llamamos a la funcion para guardar nuevos usuarios para agregar la nueva cuenta a usuarios.txt
                print(f"'{usr_new}' fue registrado con exito!\n")           # Confirmamos al usuario que su cuenta se creo correctamente

        else:
            print("\nOpción invalida.\nSeleciona una opción del 1-2\n")     # Si el usuario no ingresa un dato correcto mostramos un mensaje de error y lo regresamos al menú de inicio


def sim_ls():                                                   # Funcón que simula una pantalla de carga
    print("Cargando...\n")                                      
    for i in range(5, 0, -1):                                   # Aqui vamos a usar un ciclo for para hacer una cuenta regresiva desde 5
        print(i)            
        time.sleep(1)                                           # Usamos time.sleep(1) para ejecutar la cuenta regresiva con pausas de un segundo
    print("¡Listo!\n")                                            

def saludo(usuario):
    print(f"Hola {usuario}, ¿En qué te puedo ayudar hoy?")

def menu():                         # Función que contiene el menú de opciones
    menu = [
        ["1: Capturar ingreso"],
        ["2: Capturar egreso"],
        ["3: Imprimir resumen"],
        ["4: Imprimir reporte"],
        ["5: Salir"]
    ]
    print("")
    for i in menu:                  # Usamos un ciclo for para imprimir los indices dentro de la matriz que contiene el menú
        print(*i)                   # Usamos un print(*i) el uso del '*' es para desempaquetar los contenidos de la matriz, esto para imprimir unicamente el texto contenido 

def fecha():                                                            # Función para capturar fechas
    while True:                                                         # Abrimos ciclo while para el dia
        try:                                                            # Abrimos bloque try-except
            dia = int(input("Día"))                                     # Capturamos el dato correspondiente al día
            if dia < 1 or dia > 31:                                     # Establecemos un limite de opciones validas (1-31)
                print("Ingresa un número del 1-31")
            else:
                break                                                   # Rompemos el ciclo del día 
        except ValueError:                                              # Capturamos ValueError
            print("Ingresa los datos de la fecha en formato numérico")  

    while True:                                                         # Iniciamos ciclo para el mes
        try:    
            mes = int(input("Mes"))                                     # Capturamos el dato correspondiente al mes
            if mes < 1 or mes > 12:                                     # Establecemos limite de opciones validas (1-12)
                print("Ingresa un número del 1-12")                 
            else:
                break                                                   # Rompemos el ciclo del mes
        except ValueError:                                              # Capturamos ValueError
            print("Ingresa los datos de la fecha en formato numérico")  

    while True:                                                         # Abrimos ciclo para el año
        try:
            anio = int(input("Año"))                                    # Capturamos el dato correspondiente al año
            break
        except ValueError:                                              # Capturamos ValueError
            print("Ingresa los datos de la fecha en formato numérico") 
                                                                        # Con esta estructura de ciclos individuales para cada dato evitamos que se tenga que iniciar la captura desde el inicio en caso de que se detecte un error
    date = (dia, mes, anio)                                             # Una vez que se verifico que los datos ingresados sean correctos se guardan en una tupla con formato (dd, mm, aaaa)
    return date                                                         # La funcion retorna la tupla que contiene la fecha

def monto():                                                      # Funcion que captura montos
    while True:                                                   # abrimos ciclo de la funcion
        try:                                                      # Abrimos bloque try except
            while True:                                           # Abrimos ciclo para verificar que el dato sea mayor a 0
                monto = float(input("Ingresa el monto ($)"))      # Capturamos el monto 
                if monto <1:                                      # Establecemos una condicion para evitar que se ingresen montos de $0 o menos
                    print("Ingrese una cantidad mayor de cero")
                else:
                    break                                         # Rompemos ciclo interno
            return monto                                          
        except ValueError:                                        # Capturamos ValueError
            print("Ingrese el monto en números")

# Archivos de ingresos
with open("Ingresos.txt", mode="a") as archivo:             # Ingresos.txt solo guarda los montos de los ingresos
    archivo.write("")
with open("Resumen_ingresos.txt", mode="a") as archivo:     # Resumen_ingresos.txt guarda toda la información capturada por ingreso (fecha, monto, concepto)
    archivo.write("")

# Archivos de egresos
with open("Egresos.txt", mode="a") as archivo:              # Egresos.txt solo guarda los montos de los egresos
    archivo.write("")
with open("Resumen_egresos.txt", mode="a") as archivo:      # Resumen_egresos.txt guarda toda la información capturada por egreso (fecha, monto, concepto)
    archivo.write("")


usuario_activo = log_in()

sim_ls()                                                        # Llamamos a la función para simular la pantalla de carga 

saludo(usuario_activo)

while True: # Ciclo principal
    menu()                                                                      # Lammamos a la función que contiene el menú
    op = input(("Selecciona una opción del 1-5"))                               # Solicitamos al usuario que escoja una de las opciones del menú
    if op != "1" and op != "2" and op != "3" and op != "4" and op != "5":       # Establecemos un limite de opciones validas para el menú
        print("\n***Seleccione una opción valida***\n")
    else:    
        if op == "1":
            while True: # primer ciclo interno (ingresos)                                                                            
                print("\nINGRESOS\n")
                print("Ingresa los datos de la fecha uno por uno (dd) (mm) (aaaa)")                     # Pedimos al usuario que ingrese la fecha dato por dato
                fecha_i = fecha()                                                                       # Llamamos la funcion para capturar fechas y la asignamos a la variable fecha_i
                print(f"Fecha: {fecha_i}")          
                monto_i = monto()                                                                       # Llamamos a la función que captura los montos y la asignamos a la variable monto_i
                print(f"Ingreso: ${monto_i}")
                concept_i = input("Concepto")                                                           # Capturamos el concepto del ingreso
                print(f"Concepto: {concept_i}")

                # Guardar la información en la BD de ingresos
                with open("Ingresos.txt", mode="a") as archivo:                                         # Abrimos Ingresos.txt en modo append
                    archivo.write(f"{monto_i}\n")                                                       # Actualizamos el archivo para guardar los ingresos. Este es el archivo que usaremos para hacer la suma de los ingresos
                with open("Resumen_ingresos.txt", mode="a") as archivo:                                 # Abrimos Resumen_ingresos.txt en modo append
                    archivo.write(f"Fecha: {fecha_i} | Ingreso: ${monto_i} | Concepto: {concept_i}\n")  # Actualizamos el archivo con los datos del ingreso en formato (fecha, monto, concepto)
                print("Datos guardados correctamente\n")

                while True: #3 Ciclo para capturar más entradas
                    agregar_i = input("¿Desea agregar otro ingreso?\n 'S' para sí\n 'N' para no")       # Preguntamos al usuario si desea agregar otra entrada
                    if agregar_i.upper() != "S" and agregar_i.upper() != "N":                           # Limitamos las opciones a "s" o "n"
                        print("\n***Opción invalida***\n")
                    else:                                                                               
                        break                                                                           # Rompemos el ciclo para capturar una nueva entrada si la entrada es correcta
                if agregar_i.upper() == "N":                                                            # Si el usuario no quiere agregar mas entradas el programa regresa automaticamente al menú principal
                    break                                                                               # Rompemos el ciclo de ingresos


        if op == "2":
            while True: # Segundo ciclo interno (egresos)
                print("\nEGRESOS\n")
                print("Ingresa los datos de la fecha uno por uno (dd) (mm) (aaaa)")                     # Tambien pedimos al usuario que ingrese los datos uno por uno
                fecha_e = fecha()                                                                       # Llamamos a la funcion para capturar la fecha y ahora la asignamos a fecha_e
                print(f"Fecha: {fecha_e}")
                monto_e = monto()                                                                       # Llammos a la función para capturar montos y la asignamos a monto_e
                print(f"Egreso: ${monto_e}")
                concept_e = (input("Concepto"))                                                         # Pedimos al usuario que ingrese el concepto del egreso
                print(f"Concepto: {concept_e}")

                # Guardar la información en la BD de egresos
                with open("Egresos.txt", mode="a") as archivo:                                          # Abrimos Egresos.txt y Resumen.txt en modo append 
                    archivo.write(f"{monto_e}\n")                                                       # Actualizamos los archivo con la informacion igresada
                with open("Resumen_egresos.txt", mode="a") as archivo:
                    archivo.write(f"Fecha: {fecha_e} | Egreso: ${monto_e} | Concepto: {concept_e}\n")   # Guardamos los datos con formato (fecha, monto, concepto)
                print("Datos guardados correctamente\n")

                while True:                                                                             # Preguntamos al usuario si quiere ingresar otra entrada
                    agregar_e = input("¿Desea agregar otro egreso?\n 'S' para sí\n 'N' para no")        # El proceso es igual al de la seccion de ingresos
                    if agregar_e.upper() != "S" and agregar_e.upper() != "N":
                        print("\n***Opcion invalida***\n")
                    else:
                        break
                if agregar_e.upper() == "N":
                    break

        if op == "3":
                print("\nRESUMEN\n")
                
                with open("Resumen_ingresos.txt", mode="r") as archivo:                                             # Abrimos el Resumen_ingresos.txt en modo de lectura
                    lineas_ri = archivo.readlines()                                                                 # Leemos todas las lineas del archivo
                print("\n------------------------------------------------------------------------------------\n")
                print("Resumen de Ingresos:\n")
                for i, linea in enumerate(lineas_ri, 1):                                                            # Usamos un ciclo for y enumerate para enumerar las lines del archivo
                    print(f"{i}: {linea.strip()}")                                                                  # Imprimimos el numero de fila junto con su contenido
                print("\n------------------------------------------------------------------------------------\n")
                
                with open("Resumen_egresos.txt", mode="r") as archivo:                                              # Aqui hacemos lo mismo pero con Resumen_egresos.txt
                    lineas_re = archivo.readlines()
                print("Resumen de Egresos:\n")
                for i, linea in enumerate(lineas_re, 1):
                    print(f"{i}: {linea.strip()}")
                print("\n------------------------------------------------------------------------------------\n")
               
                
        if op == "4":
                print("\nCIERRE\n")
                with open("Ingresos.txt", mode="r") as archivo:                 # Abrimos Ingresos.txt en modo de lectura
                    ingresos = [float(linea.strip()) for linea in archivo]      # Aqui linea.strip() lee las lineas del archivo y les quita los espacios y saltos de linea, se queda solo con el texto. Convertimos los numeros guardados como texto de str a float y los asignamos a una lista (ingresos)
                total_i = sum(ingresos)                                         # Usamos sum(ingresos) para sumat todos los números que contenga la lista y guardamos el total en la variable total_i

                with open("Egresos.txt", mode="r") as archivo:                  # Aqui hacemos lo mismo que en las lineas anteriores, pero ahora con los gastos guardados en Egresos.txt
                    egresos = [float(linea.strip()) for linea in archivo]
                total_e = sum(egresos)

                with open("Reporte.txt", mode="w") as archivo:                  # Crea un archivo reporte.txt en modo de escritura
                    archivo.write(f"Total de ingresos: ${total_i}\n")           # Escribimos en el archivo el total de ingresos guardado en total_i
                    archivo.write(f"Total de egresos: ${total_e}\n")            # Escribimos en el archivo el total de egresos guardado en total_ e
                    archivo.write(f"Gran total: ${total_i - total_e}\n")        # Restamos los egresos a los ingresos y escribimos ese total en el archivo

                with open("Reporte.txt", mode="r") as archivo:                  # Abrimos Reporte.txt en modo de lectura
                    print(archivo.read())                                       # Leemos todo el archivo con .read() y lo imprimimos


        if op == "5":                       # Opción para finalizar programa
            print("\nProceso finalizado")   # mensaje de finalización
            break                           #Rompemos el ciclo principal y termina el programa
```
